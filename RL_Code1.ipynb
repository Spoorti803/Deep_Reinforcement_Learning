```python
import gymnasium as gym

# first we have to create an environmnet
env = gym.make("LunarLander-v2")
"""
now we have to reset this environmnet
env.reset returns 2 things:
observation = the current state info
info = additional info about the environment
"""
observation, info = env.reset()

#take an action 20 times
for _ in range(20):
  action = env.action_space.sample()
  print("The action taken is: ",action)


#do this action in the environment and get the observation, reward, terminated, info, truncated
"""
observation = new state
reward = reward we get after executing the action
terminated = agent reached the terminal state
truncted = time limit or the agent wnet out of the bound of environment
info = additional info about the environment
"""
observation, reward, terminated, truncated, info = env.step(action)
#if terminated or truncated then:
if terminated or truncated:
  print("Reset environmnet")
  obseravtion, info = env.reset()

env.close()
```
